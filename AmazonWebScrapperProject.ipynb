```python
#Importing libraries

from bs4 import BeautifulSoup
import requests
import smtplib
import time
import datetime
```

```python
#connecting to Amazon website

URL = 'https://www.amazon.de/-/en/Stages-Debugging-T-Shirt-Computer-Programmer/dp/B07G8Z2QBG/'

#custom user agent for your system
headers = {#Your user-agent
               }

session = requests.Session()

try:
    page = requests.get(URL, headers=headers, timeout=20)
    soup1 = BeautifulSoup(page.content, 'html.parser')

except requests.exceptions.RequestException as e:
    print('Connection Error: ', e)

```

```python
soup2 = BeautifulSoup(soup1.prettify(), "html.parser")

title = soup2.find(id='productTitle').get_text(strip=True)
price = soup2.find(class_ = "aok-offscreen").get_text(strip=True)[1:]
color = soup2.find(id="inline-twister-expanded-dimension-text-color_name").get_text(strip=True)
today = datetime.date.today()

print(title)
print(price)
print(color)
print(today)
```

```python
#creating csv file 

import csv

header = ['Title','Price','Color','Date']
data = [title, price, color,today]

with open('AmazonWebScraperDataset.csv', 'w', newline='', encoding='UTF-8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerow(data)
```
